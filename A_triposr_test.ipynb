```python
from google.colab import userdata
import subprocess, os

GH_USER = "ekatet19915-cpu"
REPO    = "3D_model"
NAME    = "Kate"
EMAIL   = "ekatet19915@gmail.com"

token = userdata.get("GH_TOKEN").strip()
REPO_DIR = f"/content/{REPO}"

def run(cmd, show=True):
    r = subprocess.run(cmd, shell=True, capture_output=True, text=True)
    out = (r.stdout + r.stderr).replace(token, "***")
    if show and out.strip():
        print(f"$ {cmd.replace(token, '***')}\n{out}")
    return r

if not os.path.exists(REPO_DIR):
    r = run(f"git clone https://{token}@github.com/{GH_USER}/{REPO}.git {REPO_DIR}")
    if r.returncode != 0:
        raise RuntimeError("Clone failed, see message above")

os.chdir(REPO_DIR)
run(f'git config user.name "{NAME}"', show=False)
run(f'git config user.email "{EMAIL}"', show=False)

def sync(msg="Update from Colab"):
    """Pull others' changes, then commit and push ours."""
    os.chdir(REPO_DIR)
    run("git pull --rebase --autostash")
    run("git add -A")
    run(f'git commit -m "{msg}"')
    run("git push")

print("Ready. Working folder:", os.getcwd())
```

```python
shim = '''
import numpy as np, torch
from skimage.measure import marching_cubes as _mc

def marching_cubes(vol, thresh):
    """Drop-in replacement for torchmcubes.marching_cubes (CPU, scikit-image)."""
    v = vol.detach().cpu().numpy().astype(np.float32)
    verts, faces, _, _ = _mc(v, level=float(thresh))
    verts = np.ascontiguousarray(verts[:, [2, 1, 0]])
    faces = np.ascontiguousarray(faces.astype(np.int64))
    return torch.from_numpy(verts).float(), torch.from_numpy(faces)
'''
with open("/content/TripoSR/torchmcubes.py", "w") as f:
    f.write(shim)
print("ok")
```
