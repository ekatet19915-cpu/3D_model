```python
import os
import shutil
from google.colab import files

# Nastavte cestu k cieľovému priečinku
target_dir = "/content/stable-fast-3d"
os.makedirs(target_dir, exist_ok=True)

print("Pretiahnite fotku sem (alebo kliknite na 'Choose Files'):")
uploaded = files.upload()

# Presun nahraných súborov do cieľového priečinka
for filename in uploaded.keys():
    dest_path = os.path.join(target_dir, filename)
    shutil.move(filename, dest_path)
    print(f"Fotka uložená do: {dest_path}")
```

```python
%%shell
eval "$(conda shell.bash hook)"
conda activate sf3d_env

cd /content/stable-fast-3d

# Vytvorenie patched skriptu run_nomath.py
cat << 'EOF' > run_nomath.py
import runpy, torch
import torch.nn.functional as F
from torch.nn.attention import sdpa_kernel, SDPBackend

_orig = F.scaled_dot_product_attention

def _patched(q, k, v, *args, **kwargs):
    if q.is_cuda and q.dtype == torch.bfloat16:
        dt = q.dtype
        q, k, v = q.half(), k.half(), v.half()
        m = kwargs.get("attn_mask")
        if m is not None and m.dtype == torch.bfloat16:
            kwargs["attn_mask"] = m.half()
        try:
            with sdpa_kernel([SDPBackend.EFFICIENT_ATTENTION]):
                return _orig(q, k, v, *args, **kwargs).to(dt)
        except RuntimeError:
            outs = []
            for i in range(0, q.shape[2], 1024):
                outs.append(_orig(q[:, :, i:i+1024], k, v, *args, **kwargs))
            return torch.cat(outs, dim=2).to(dt)
    return _orig(q, k, v, *args, **kwargs)

F.scaled_dot_product_attention = _patched
runpy.run_path("run.py", run_name="__main__")
EOF

# Spustenie generovania s ošetreným skriptom
export PYTORCH_CUDA_ALLOC_CONF="expandable_segments:True"
python run_nomath.py /content/stable-fast-3d/muz.png \
  --output-dir /content/output \
  --texture-resolution 1024 \
  --remesh_option triangle \
  --target_vertex_count 25000
```
