```python
import re, subprocess, os

REPO = "/content/stable-fast-3d"
skip = {"torch", "torchvision", "numpy", "transformers", "pillow", "trimesh",
        "rembg", "huggingface-hub", "huggingface_hub", "einops", "omegaconf",
        "setuptools", "onnxruntime"}

to_install = []
for line in open(f"{REPO}/requirements.txt"):
    line = line.strip()
    if not line or line.startswith("#"):
        continue
    if line.startswith((".", "git+")):          # local folders / git packages: keep as is
        to_install.append(line)
        continue
    name = re.split(r"[=<>\[ ;]", line)[0].lower()
    if name in skip:
        continue
    to_install.append(name)                      # drop the version pin

print("Will install:", to_install)
for pkg in to_install:
    r = subprocess.run(f"cd {REPO} && pip install {pkg}", shell=True,
                       capture_output=True, text=True)
    print("OK  " if r.returncode == 0 else "FAIL", pkg)
    if r.returncode != 0:
        print(r.stderr[-800:])
```

```python
import glob, os, re, time, csv, shutil, subprocess

PHOTOS = sorted(glob.glob("/content/3D_model/data/test/*.jpg"))
OUT_DIR = "/content/sf3d_out"
os.makedirs(OUT_DIR, exist_ok=True)

# simple_01 was processed earlier into sf3d_out/0 -> copy to a named folder
if os.path.exists(f"{OUT_DIR}/0/mesh.glb") and not os.path.exists(f"{OUT_DIR}/simple_01"):
    shutil.copytree(f"{OUT_DIR}/0", f"{OUT_DIR}/simple_01/0")

env = dict(os.environ, PYTORCH_CUDA_ALLOC_CONF="expandable_segments:True")
rows = []
os.chdir("/content/stable-fast-3d")
for p in PHOTOS:
    name = os.path.splitext(os.path.basename(p))[0]
    glb = f"{OUT_DIR}/{name}/0/mesh.glb"
    if os.path.exists(glb):
        print(name, "- already done, skipping")
        continue
    t = time.time()
    r = subprocess.run(["python", "run.py", p, "--output-dir", f"{OUT_DIR}/{name}"],
                       capture_output=True, text=True, env=env)
    total = round(time.time() - t, 1)
    ok = os.path.exists(glb)
    log = r.stdout + r.stderr
    m = re.search(r"Peak Memory: ([\d.]+)", log)
    peak = round(float(m.group(1))) if m else ""
    rows.append([name, "ok" if ok else "FAILED", total, peak])
    print(name, "ok" if ok else "FAILED", f"{total}s")
    if not ok:
        print("   ", log.strip().splitlines()[-1][:200])

os.makedirs("/content/3D_model/docs", exist_ok=True)
csv_path = "/content/3D_model/docs/sf3d_timing.csv"
new_file = not os.path.exists(csv_path)
with open(csv_path, "a", newline="", encoding="utf-8") as f:
    w = csv.writer(f)
    if new_file:
        w.writerow(["file", "status", "total_s", "peak_mem_mb"])
    w.writerows(rows)
print("\nTable saved:", csv_path)
```
